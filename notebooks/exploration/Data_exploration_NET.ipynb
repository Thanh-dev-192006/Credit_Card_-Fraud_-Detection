```python
column_change = [c for c in df.columns if c not in ['Weight', "fraud"]]
column_change
```

```python
df[column_change] = df[column_change].replace("'", "", regex=True)
```

```python
df.head()
```

```python
df_normal = df[df['fraud'] == 0]['Weight']
df_fraud = df[df['fraud'] == 1]['Weight']
statistics = pd.DataFrame({
    "Stats": ['Mean', 'Median', 'Max', 'Min', "Std"],
    "Normal": [df_normal.mean(), df_normal.median(), df_normal.max(), df_normal.min(), df_normal.std()],
    "Fraud": [df_fraud.mean(), df_fraud.median(), df_fraud.max(), df_fraud.min(), df_fraud.std()]
})

print(statistics)
```
